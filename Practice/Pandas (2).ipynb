```python
#I want to display the avg sales and profit  for each region in same table?
data.groupby("Region")[["Sales","Profit"]].mean()
```

```python
#I want to display the avg and sum of sales and profit  for each region in same table?
import numpy as np
data.groupby("Region").agg({"Sales":["mean","sum"],"Profit":["mean","sum","std"]})
```

```python
# dispaly the teh avg profit for the deliveries happened through different shipment mode in different region?
data.groupby(["Ship Mode","Region"])["Profit"].mean().unstack()
```
