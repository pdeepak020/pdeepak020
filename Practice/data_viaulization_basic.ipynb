```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
t=np.linspace(0,20,10)
t
```

```python
y=np.sin(t)
y
```

```python
#lineplot
plt.plot(t,y)
plt.show()
```

```python
y1=np.cos(t)
plt.plot(t,y1)
plt.show()
```

```python
plt.plot(t,y,color="r",marker="*",label="sinwave")
plt.plot(t,y1,color="g",label="coswave")
plt.legend()# to print the label
plt.show()
```

```python
#sactterplot
plt.scatter(t,y)
plt.show()
```

```python
plt.bar(t,y)
plt.show()
```

```python
#Barchart------->bivariant plot (1 numeric, 1 category)
x=["A","B","C","D"]
y=[24,67,12,89]
plt.bar(x,y)
plt.show()
```
