```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
# import warnings
# warnings.filterwarnings("ignore")
```

```python
t=np.linspace(0,20,10)
t
```

```python
y=np.sin(t)
y
```

```python
#lineplot
plt.plot(t,y)
plt.show()
```

```python
y1=np.cos(t)
plt.plot(t,y1)
plt.show()
```

```python
y2=np.tan(t)
plt.plot(t,y2)
plt.show()
```

```python
plt.plot(t,y,color="r",marker="*",label="sinwave")
plt.plot(t,y1,color="g",label="coswave")
plt.plot(t,y2,color="m",label="tanwave")
plt.legend()# to print the label
plt.show()
```

```python
# subplot
plt.figure(figsize=(10,8))
plt.subplot(2,2,1)
plt.plot(t,y,color="r",marker="*",label="sinwave")
plt.legend()

plt.subplot(2,2,2)
plt.plot(t,y1,color="g",label="coswave")
plt.legend()


plt.subplot(2,2,3)
plt.plot(t,y2,color="m",label="tanwave")
plt.legend()


plt.show()

```

```python
#sactterplot
plt.scatter(t,y)
plt.show()
```

```python
plt.bar(t,y)
plt.show()
```

```python
#Barchart------->bivariant plot (1 numeric, 1 category)
x=["A","B","C","D"]
y=[24,67,12,89]
plt.bar(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.show()
```

```python
#histogram(data distribution)-------->univariant

data=np.random.randn(1000)
plt.hist(data)
plt.show()
```

```python
# piechart------->bivariant ( 1numerical and 1 categorical)# percentage distribution
agent=["John","Kumar","ben","Dev","giri"]
loan=[24,18,29,7,32]
#plt.bar(agent,loan)
plt.pie(loan,labels=agent,autopct="%.1f%%",explode=[0,0,0,0,0.3])
plt.show()
```

# seaborn

```python
import seaborn as sns
#sns.get_dataset_names() get different datasets in seaborn
```

```python
data1=sns.load_dataset('tips')
data1
```

```python
data1.info()
```

```python
#find the relation between total bill and tips
plt.figure(figsize=(10,5))
sns.lineplot(x=data1["total_bill"],y=data1["tip"])
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x=data1["total_bill"],y=data1["tip"],hue=data1["sex"])
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x=data1["total_bill"],y=data1["tip"],hue=data1["time"])
plt.show()
```

```python
#barplot(1 numerical,categorical)
# what is the max amount of bills paid by male and female
sns.barplot(x=data1["sex"],y=data1["total_bill"],ci=None,estimator=np.max,hue=data1["time"])
plt.show()
```

```python
data1.groupby("sex")["total_bill"].mean()
```

```python
sns.barplot(x=data1["time"],y=data1["tip"],ci=None,estimator=np.max,hue=data1["day"])
plt.show()
```

```python
data1["day"].value_counts()
```

```python
sns.countplot(x=data1["day"])
plt.show()
```

```python
#statistical plot

# Heatmap--------->correlation analysis(multivariant)

# boxplot-----------> outlier detection

# Violin plot---------->another representaion of outlier treatment
```

```python
# correlation analysis(only for numerical)

data1.corr(numeric_only=True)
```

```python
sns.heatmap(data1.corr(),annot=True,cmap="coolwarm")
plt.show()
```

```python
# boxplot(outlier detection)
sns.boxplot(x=data1["total_bill"])
plt.show()
```
