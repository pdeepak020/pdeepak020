```python
ypred1=180+1.8*df1["age"]
ypred2=150+0.75*df1["age"]
```

```python
np.sum((df1["glucose"]-ypred1)**2)
```

```python
np.sum((df1["glucose"]-ypred2)**2)
```

```python
# formula based approach
x=df1["age"]
y=df1["glucose"]
```

```python
#formula
# b1=sum((x-x1)*(y-y1))/sum ((x-x1)^2)
# b0=y1-b1*x1
```

```python
beta1=np.sum((x-np.mean(x))*(y-np.mean(y)))/ np.sum((x-np.mean(x))**2)
beta1
```

```python
beta0=np.mean(y)-beta1*np.mean(x)
beta0
```

```python
ybest=beta0+beta1*x
ybest
```

```python
plt.plot(df1["age"],df1["glucose"],"*")
plt.plot(df1["age"],yprednull)
plt.plot(df1["age"],ypred1)
plt.plot(df1["age"],ypred2)
plt.plot(df1["age"],ybest)
#plt.legend(loc=4)
```

```python
np.sum(y-ybest)
```

```python
np.sum((y-ybest)**2)
```

```python
np.sqrt(np.mean((y-ybest)**2))
```
