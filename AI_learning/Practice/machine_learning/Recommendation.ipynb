### Popularity Based Recommendation System

```python
# Top 10 books interms of average rating 

pd.DataFrame(data.groupby('bookTitle')['bookRating'].mean().sort_values(ascending=False)).head(10)
```

```python
popularity_table=data.groupby('bookTitle').agg({'bookRating':'mean','totalRatingCount':'count'})
```

```python
popularity_table
```

```python
#consider the books for recommendation only if it has 100 rating counts
top_popularity_table=popularity_table[popularity_table['totalRatingCount']>100]
top_popularity_table
```

```python
#Top 10 books to recommed based on popularity
top_popularity_table.sort_values('bookRating',ascending=False).head(10)
```

```python
#Top 10 books to recommed based on popularity
top_popularity_table.sort_values('totalRatingCount',ascending=False).head(10)
```

###  Content-based recommendation system

```python
data1=pd.read_csv('movie_metadata.csv')
```

```python
data1.head(2)
```

```python
data1.columns
```

```python
data1["genres"]
```

```python
genres=data1['genres'].str.split('|',expand=True)
#genere=data1["genres"].str.split('|')
#genres
genres=genres.iloc[:,0:3]
```

```python
genres
```

```python
genres=genres.fillna('Others')
```

```python
genres
```

```python
genres.columns=['genre1','genre2','genre3']
genres.head(10)
```

```python
data1=pd.concat([data1,genres],axis=1)
```

```python
data1.head()
```

```python
#movie_feat=['movie_title','genre1','genre2','genre3','language','content_rating',
           #'imdb_score','num_user_for_reviews','movie_facebook_likes']
```

```python
movie_feat=['movie_title','genre1','genre2','genre3','content_rating','imdb_score']
```

```python
data2=data1[movie_feat]
```

```python
data2.head(2)
```

```python
data2.shape
```

```python
data2=data2.drop_duplicates()
```

```python
data2.index=data2['movie_title']
```

```python
data2
```

```python
data2=data2.drop('movie_title',axis=1)
```

```python
data2.head(2)
```

```python
data3=pd.get_dummies(data2)# one hot encoding
```

```python
data3.head()
```

```python
data3.shape
```

```python
data3=data3.dropna()
data3
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
rec_model = NearestNeighbors(metric = 'cosine')
rec_model.fit(data3)
```

```python
data3.index
```

```python
#data4=data3
data3.index=data3.index.str.strip()
```

```python
query_movie="X-Men 2"
data3.loc[query_movie, :].values.reshape(1, -1)
```

```python
query_movie='Spectre'
#query_movie="Licence to Kill"
dist, ind = rec_model.kneighbors(data3.loc[query_movie, :].values.reshape(1, -1), n_neighbors = 6)
```
