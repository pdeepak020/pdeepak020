# Dynamic Message Generation in LangChain

In this section, we demonstrate how to use **LangChain's prompt templates** to dynamically insert variables and past conversation history.

```python
from langchain.prompts import MessagesPlaceholder, HumanMessagePromptTemplate, ChatPromptTemplate
from langchain.schema import HumanMessage

# Step 1: Define a simple template with placeholders
dynamic_prompt = "The {skill} is essential for success in {domain}."

# Step 2: Convert to a LangChain message template
message_template = HumanMessagePromptTemplate.from_template(dynamic_prompt)

# Step 3: Define a user message
human_message = HumanMessage(content="How can I improve my data science skills?")

# Step 4: Create a chat prompt combining system + template + conversation history
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    message_template,
    MessagesPlaceholder(variable_name="conversation_history")
])

# Step 5: Format the conversation with values
conversation = chat_prompt.format_prompt(
    skill="Critical Thinking",
    domain="Data Science",
    conversation_history=[human_message]
)

# Step 6: Print the conversation
for message in conversation.to_messages():
    print(message)
```

```python
from langchain.schema import AIMessage

# Step 7: Add an AI response to the conversation history
ai_message = AIMessage(content=(
    "To improve your data science skills, practice real-world projects, "
    "strengthen your math foundations, and stay updated with new tools."
))

# Update the conversation history with both Human and AI messages
conversation = chat_prompt.format_prompt(
    skill="Critical Thinking",
    domain="Data Science",
    conversation_history=[human_message, ai_message]
)

# Print updated conversation
for message in conversation.to_messages():
    print(message.content)
```
