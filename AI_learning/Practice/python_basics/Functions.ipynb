```python
# Arbitary parameter------------>variable length paarmeter
def additionnum(*args):
    tot=0
    for i in args:
        tot=tot+i
    return tot
    
    
```

```python
additionnum(10,20,30,40,50,90,150,-8)
```

# File handling

```python
# Read the file

with open("demo.txt","r") as raja:
    content=raja.read()
    print(content)
    
```

```python
# write tha data into the file(overwrite)
with open("demo.txt","w") as file:
    file.write("Hello world!\n")
    file.write("America\n")
               
```

```python
# write tha data into the file(overwrite)
with open("demo.txt","a") as file:
    file.write("Appending the new line\n")
    file.write("line 2\n")
```
