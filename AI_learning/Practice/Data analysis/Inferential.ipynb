```python
#classical
trial=[]
for i in np.arange(100):    
    samp=[]
    for val in np.arange(500):
        samp.append(np.random.choice(w_pop))
    trial.append(np.mean(samp))

    
```

```python
len(trial)
```

```python
sns.distplot(trial)
```

```python
import pandas as pd
pd.Series(trial).skew()
```

```python
me=np.array(trial).mean()
me
```

```python
s_d=np.array(trial).std()
s_d
```

```python
me-3*(s_d),me+3*(s_d) # 99.7 percentage of data lie in this limit
```

```python
me-2*(s_d),me+2*(s_d) # 95.5 percentage of data lie in this limit
```

```python
#What is the data range for 95 percent of data ?
me-1.96*(s_d),me+1.96*(s_d) # 95 percentage of data lie in this limit
```

```python
# Method 2-----Practical
ss=[]
for i in np.arange(500):
    ss.append(np.random.choice(w_pop))
```

```python
len(ss)
```

```python
x_bar=np.mean(ss)
x_bar
```

```python
#increase the sample size reduce the sample mean error
std_err=np.std(ss)/np.sqrt(500)
```

```python
#What is the data range for 95 percent of data ?
x_bar-1.96*std_err,x_bar+1.96*std_err
```
